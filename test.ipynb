```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error, r2_score
import pandas as pd

df_train = pd.read_csv("./trash/data/train_new.csv")
df_train = df_train.drop(columns="ChemID")
# df_train = pd.read_csv("./trash/data/train_new.csv")
# df_test = pd.read_csv("./src/data/test.csv")


X = df_train.drop(columns="LogP")
y = df_train["LogP"]

X = X[["MW", "nHBDon", "nHBAcc", "nRing", "TopoPSA(NO)", "nAromAtom", "Zagreb1", "SLogP", "BalabanJ", "Kier1"]]
```

```python

model = RandomForestRegressor(n_estimators=1000, max_depth=None,
                                      min_samples_split=3, random_state=42)
model.fit(X, y)
```

```python
y_pred = model.predict(X)
r_squared = r2_score(y, y_pred)
print("training r_squared score : ")
print(r_squared)
print(f"RMSE : {root_mean_squared_error(y, y_pred)}")

print("===================================")
```
